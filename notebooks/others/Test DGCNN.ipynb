```python
import torch
from torch.utils.data import DataLoader

from torch_pointcloud.models.dgcnn import DGCNN
from torch_pointcloud.datasets.modelnet import ModelNet10
import torch_pointcloud.transforms as T
```

```python
# Dataset
pre_transform = T.NormalizeScale()
transform = T.Compose([T.SampleMeshPoints(1024)])

dataset = ModelNet10(root="../data", train=False, pre_transform=pre_transform, transform=transform)


# DataLoader
def collate(data_list):
    return {
        "pos": torch.stack([d["pos"] for d in data_list]),
        "target": torch.cat([d["target"] for d in data_list]),
    }
        
dataloader = DataLoader(dataset, batch_size=32, shuffle=True, collate_fn=collate)
for data in dataloader:
    break
```

```python
model = DGCNN(num_classes=10).cuda()
```

```python
%%time
model(data["pos"].permute(0, 2, 1).cuda())
```
